```python
import pandas as pd
import numpy as np

np.random.seed(10)
df = pd.DataFrame({
    "Year": np.arange(2015, 2025),
    "Sales": np.random.randint(200,600,10),
    "Profit": np.random.randint(50,200,10),
    "Customers": np.random.randint(1000,5000,10),
})
```

```python
import matplotlib.pyplot as plt
base_path ="/home/user/Desktop/24-03-2026/"
print(base_path)

plt.figure()
plt.plot(df["Year"], df["Sales"], marker='o')
plt.xlabel("Year")
plt.ylabel("Sales")
plt.title("Sales trend over time")

p = base_path+"line_plot.png"
plt.savefig(p)
plt.show()
plt.close()

print(p)
```

```python
import matplotlib.pyplot as plt
base_path ="/home/user/Desktop/24-03-2026/"
print(base_path)

plt.figure()
plt.bar(df["Year"].astype(str), df["Profit"])
plt.xlabel("Year")
plt.ylabel("Profit")
plt.title("Profit by year")

p = base_path+"bar_chart.png"
plt.savefig(p)
plt.show()
plt.close()

print(p)
```

```python
import matplotlib.pyplot as plt
base_path ="/home/user/Desktop/24-03-2026/"
print(base_path)

plt.figure()
plt.scatter(df["Customers"], df["Sales"])
plt.xlabel("Customers")
plt.ylabel("Sales")
plt.title("Customers VS Sales")

p = base_path+"scatter_plot.png"
plt.savefig(p)
plt.show()
plt.close()

print(p)
```

```python
df = sns.load_dataset('car_crashes')
tips = sns.load_dataset('tips')
iris= sns.load_dataset('iris')
help(iris)

print(df.head())
print(df.tail())
```

```python
print("\nSUMMARY STATISTICS")
print(iris.describe())

print("\n DATA TYPES")
print(iris.dtypes)
numeric_iris = iris.select_dtypes(include = 'number')

print("\n CORRELAION MATRIX")
corr = numeric_iris.corr()
print(corr)

print("\n COVARIANCE MATRIX")
cov = numeric_iris.cov()
print(cov)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

base_path ="/home/user/Desktop/24-03-2026/"

plt.figure(figsize=(10,6))
sns.heatmap(corr, annot = True, cmap='viridis', fmt='.2f')
plt.title("Correlation Heatmap")
p = base_path+"corr_heatmap.png"
plt.savefig(p)
plt.show()
plt.close()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

base_path ="/home/user/Desktop/24-03-2026/"

plt.figure(figsize=(10,6))
sns.heatmap(cov, annot = True, cmap='viridis', fmt='.2f')

plt.title("Covirance Heatmap")
p = base_path+"cov_heatmap.png"
plt.savefig(p)
plt.show()
plt.close()
```
