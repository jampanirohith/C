```python
import csv

with open('/home/user/Desktop/17-03-2026/data1.csv', 'w') as file:
    write = csv.writer(file)
    write.writerow(['Name', 'Class'])
    write.writerow(['Ram', 'xi'])
    write.writerow(['Ajith', 'xii'])

with open('/home/user/Desktop/17-03-2026/data1.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row)
```

```python
import csv

with open('/home/user/Desktop/17-03-2026/data2.csv', 'w') as file:
    write = csv.writer(file)
    feild = ['Name', 'Class']
    rows = [['Ram', 'xi'], ['Ajith','xii']]
    write.writerow(feild)
    write.writerow(rows)

with open('/home/user/Desktop/17-03-2026/data2.csv', 'r') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(dict(row))
```
