```python
import pandas as pd

data = pd.Series([10, 20, 30, 40, 50])

print("mean:", data.mean())
print("Standard Deviation:", data.std())
data = data*2
print("Updated Series:")
print(data)
```

```python
import pandas as pd

data = {
    'apples':[3,2,0,1],
    'oranges':[0,3,7,2]
}

purchases = pd.DataFrame(data, index=['June', 'Robert', 'Lily', 'David'])

print(purchases)

print(purchases.loc['June'])
print(purchases.iloc[0])
```

```python
import pandas as pd
data={
    'Name': ['A', 'B', 'C'],
    'Marks': [60, 75, 90]
}
df=pd.DataFrame(data)
df['Percentage']=df['Marks']/100*100
avg = df['Marks'].mean()
print("Strudents aboveaverage:")
print(df[df['Marks'] > avg]['Name'])
```
